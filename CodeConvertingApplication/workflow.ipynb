```python
from graphviz import Digraph

dot = Digraph(comment='Code Conversion Workflow')

# Nodes
dot.node('A', 'User Login')
dot.node('B', 'Conversion Interface\n- Input source code\n- Choose languages')
dot.node('C', 'Frontend sends POST /convert\n(user_id, code, languages)')
dot.node('D', 'Flask Backend\n- Call OpenAI API\n- Receive converted code')
dot.node('E', 'Call Oracle Procedure StoreCodeConversion\n- Check source code validity\n- Check daily conversion limit\n- Insert conversion record\n- Log conversion history')
dot.node('F', 'Error')
dot.node('G', 'Success')
dot.node('H', 'Backend returns error')
dot.node('I', 'Backend returns success')
dot.node('J', 'Frontend displays error')
dot.node('K', 'Frontend displays converted code and status')
dot.node('L', 'Update conversion count')
dot.node('M', 'User views conversion history')

# Edges
dot.edge('A', 'B')
dot.edge('B', 'C')
dot.edge('C', 'D')
dot.edge('D', 'E')
dot.edge('E', 'F', label='If error')
dot.edge('E', 'G', label='If success')
dot.edge('F', 'H')
dot.edge('G', 'I')
dot.edge('H', 'J')
dot.edge('I', 'K')
dot.edge('K', 'L')
dot.edge('L', 'M')

# Render diagram to file (PDF or PNG)
dot.render('code_conversion_workflow', format='png', view=True)  # Sẽ tạo file code_conversion_workflow.png và mở nó

# Nếu không muốn tự động mở file, dùng: dot.render('code_conversion_workflow', format='png')
```
